```python
# For upcoming assingments, START with two layers like in the original examples and change
# the layers based on questions. 2 & 4 cannot be correctly answered if you go with one only
import tensorflow as tf
print(tf.__version__)
mnist = tf.keras.datasets.mnist
(training_images, training_labels), (test_images, test_labels) = mnist.load_data()
training_images=training_images.reshape(60000, 28, 28, 1)
training_images=training_images / 255.0
test_images = test_images.reshape(10000, 28, 28, 1)
test_images=test_images/255.0
model = tf.keras.models.Sequential([
  tf.keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=(28, 28, 1)), # 32 Convolutions, # 1
  tf.keras.layers.MaxPooling2D(2, 2),
  tf.keras.layers.Flatten(),
  tf.keras.layers.Dense(128, activation='relu'),
  tf.keras.layers.Dense(10, activation='softmax')
])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(training_images, training_labels, epochs=10)
test_loss, test_acc = model.evaluate(test_images, test_labels)
print(test_acc)
# 1 convolution implementations, 32 convolutions for the set
# Final Accuracy: 98%, time taken to complete: 57 seconds
```

```python
# CNN version of code with loss function implementation as
# asked of question 5

import time
import tensorflow as tf
print(tf.__version__)

class StopAtLoss(tf.keras.callbacks.Callback):
    def __init__(self, threshold=0.01):
        super().__init__()
        self.threshold = threshold # threshold

    def on_epoch_end(self, epoch, logs=None):
        logs = logs or {}
        if logs.get('loss') is not None and logs['loss'] < self.threshold:
            print(f"\nLoss {logs['loss']:.4f} is below {self.threshold}, stopping training at epoch {epoch + 1}.")
            self.model.stop_training = True # stop model training once a certain threshold is hit

mnist = tf.keras.datasets.mnist
(training_images, training_labels), (test_images, test_labels) = mnist.load_data()
training_images = training_images.reshape(60000, 28, 28, 1) / 255.0
test_images = test_images.reshape(10000, 28, 28, 1) / 255.0

model = tf.keras.models.Sequential([
  tf.keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=(28, 28, 1)),
  tf.keras.layers.MaxPooling2D(2, 2),
  tf.keras.layers.Conv2D(32, (3,3), activation='relu'),
  tf.keras.layers.MaxPooling2D(2, 2),
  tf.keras.layers.Flatten(),
  tf.keras.layers.Dense(128, activation='relu'),
  tf.keras.layers.Dense(10, activation='softmax')
])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

start = time.time()
model.fit(training_images, training_labels, epochs=10, callbacks=[StopAtLoss(0.01)])
print(f"Training time: {time.time() - start:.1f} s")

test_loss, test_acc = model.evaluate(test_images, test_labels)
print(test_acc)
# Two convolutions, at 32 convolutions
# Function is supposed to activate when accuracy falls below a certain threshold
```
